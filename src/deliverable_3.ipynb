## Classes overview

### Character

The character class build different characters to fight against each other. 

The class has the following attributes:
* name to create a name for the character
* hp_max to define a maximum health-points
* damage to determine how much damage the character make
* weapon to define if the character has a weapon to use in battles

The class has the following methods:
* take_damage() which reduces a characters HP
* heal() which increase a characters HP
* attack() which uses take_damage() to target other characters, and can use weapons to deal damage
* introduce() which creates a string to introduce the character

The character class has following two properties:
* hp() which is used to inspect a characters HP whitout changing it
* is_alive() which is used to check if the character is alive

The character class has two subclasses:
* Warrior: can use an armor (new attribute) to protect himself, the take_damage() method is changed to that the damage on the warrior is smaller than on a normal character. 
* Mage: can use magic by using mana (new attribute) to attack with double damage, the attack() method is changed so that it is possible to use mana to create a more powerful attack.

### Dice

* Class dice has attributes sides used to determine the highest number the dice can roll.
* Dice has method roll that generates a random number between 1 and sides.
* \_\_call\_\_ uses roll to roll a dice object created with the class.

### Items

Weapon attributes:

* Name: the name of the weapon
* Damage dice: the size/number of sides of the dice that is rolled when dealing damage with the weapon.

Weapon methods:

* Roll_damage: determines how much damage a single hit from the weapon does by rolling the weapon dice

### Battle

The battle class is for simulating battles between two characters generated with the Character class. Attributes:

* hero: the hero character in the battle, always goes first and will be celebrated if they win the battle.
* enemy: the enemy character in the battle, always goes last and will be blamed if they win the battle.

Methods:

* play_round: advances a single round of combat. Each character attacks once, starting with the hero character. If the enemy character is still alive after the heroes attack, they will attack.
* simulate_battle: calls play_round until either character is no longer alive.

## Simulated Battles

```python
aria = game.Character("Aria", weapon=items.Weapon("Short Sword", Dice(20)))
wolf = game.Character("Wolf", damage=15)
goblin = game.Character("Goblin", hp_max=200, weapon=items.Weapon("Dagger", Dice(12)))
slime = game.Character("slime")

warrior = game.Warrior("Warrior", hp_max=150, weapon=items.Weapon("Ax", Dice(25)))
mage = game.Mage("Mage", weapon=items.Weapon("Staff", Dice(25)))

print(Battle(aria, wolf).simulate_battle())
print(Battle(warrior, goblin).simulate_battle())
print(Battle(mage, slime).simulate_battle())
```
