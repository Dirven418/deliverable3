## Classes overview

### Character

The character class build different characters to fight against each other. 

The class has the following attributes:
* name to create a name for the character
* hp_max to define a maximum health-points
* damage to determine how much damage the character make
* weapon to define if the character has a weapon to use in battles

The class has the following methods:
* take_damage() which reduces a characters HP
* heal() which increase a characters HP
* attack() which uses take_damage() to target other characters, and can use weapons to deal damage
* introduce() which creates a string to introduce the character

The character class has following two properties:
* hp() which is used to inspect a characters HP whitout changing it
* is_alive() which is used to check if the character is alive

The character class has two subclasses:
* Warrior: can use an armor (new attribute) to protect himself, the take_damage() method is changed to that the damage on the warrior is smaller than on a normal character. 
* Mage: can use magic by using mana (new attribute) to attack with double damage, the attack() method is changed so that it is possible to use mana to create a more powerful attack.
